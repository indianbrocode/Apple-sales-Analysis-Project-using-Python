## **Top 10 highest rated IPhone on flipkart in India**

```python
Highest_Rated=Data.sort_values(by=["Star Rating"],ascending=False)

Highest_Rated=Highest_Rated.head(10)
```

```python
print(Highest_Rated["Product Name"])
```

```python
print(Highest_Rated)
```

## **2.How many rating so the higehest-rated iPhones on Flipkart have?**

```python
iPhones=Highest_Rated["Product Name"].value_counts()
label=iPhones.index
counts=Highest_Rated["Number Of Ratings"]
figure=px.bar(Highest_Rated,x=label, y=counts)
figure.show()
```

```python
Data.head()
```

## **Which iPhone has the highest number of review onn flipkart**

```python
iPhones=Highest_Rated["Product Name"].value_counts()
label=iPhones.index
counts=Highest_Rated["Number Of Reviews"]

figure=px.bar(Highest_Rated,x=label, y=counts)
figure.show()
```

# **What is the relationship between the Sales price of the iPhone and the number of rating on the flipkart**

```python
Data.head()
```

```python
figure=px.scatter(data_frame=Data, x="Number Of Ratings",y="Sale Price",size="Discount Percentage",trendline='ols',title="Sales price")
figure.show()
```
